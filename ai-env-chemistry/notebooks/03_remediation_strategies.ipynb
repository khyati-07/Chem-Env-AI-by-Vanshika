## Water Quality: Contamination Rate Analysis

```python
water_df = load_sample_water_quality()

# Per-site contamination rate
site_stats = water_df.groupby('site')['contaminated'].agg(['sum', 'count'])
site_stats['rate_pct'] = (site_stats['sum'] / site_stats['count'] * 100).round(1)
print(site_stats)

fig, ax = plt.subplots(figsize=(7, 4))
site_stats['rate_pct'].plot(kind='bar', color='tomato', ax=ax)
ax.set_title('Contamination Rate by Site (%)')
ax.set_ylabel('Contamination Rate %')
ax.set_xlabel('Site')
plt.tight_layout()
plt.show()
```
